Parte B – medidas de pressão arterial

```python
#4 B.01 ------------------------------------------------------------------------------------------

# Importa a biblioteca de funções matemáticas da linguagem Python.
import math

# Importa a biblioteca NumPy e cria o apelido convencional np.
import numpy as np

# Importa o módulo de gráficos pyplot e cria o apelido plt.
import matplotlib.pyplot as plt

#valores originais
pressao_arterial = np.array([121.7, 120.4, 122.1, 119.8, 121.2, 120.9, 122.4, 121.0, 120.2, 121.5])
pa_media = np.mean(pressao_arterial)
pa_min = min(pressao_arterial)
pa_max = max(pressao_arterial)
pa_desvio_padrao = np.std(pressao_arterial)
pa_amplitude = pa_max - pa_min
print(f"Valor Original - Media: {pa_media}")
print(f"Valor Original - Mínimo: {pa_min}")
print(f"Valor Original - Máximo: {pa_max}")
print(f"Valor Original - Desvio Padrão: {pa_desvio_padrao}")
print(f"Valor Original - Amplitude: {pa_amplitude}")
print("\n")

#valores arredondados para o inteiro mais proximo
pressao_arterial_int = np.round(pressao_arterial)
pa_media_int = np.mean(pressao_arterial_int)
pa_min_int = min(pressao_arterial_int)
pa_max_int = max(pressao_arterial_int)
pa_desvio_padrao_int = np.std(pressao_arterial_int)
pa_amplitude_int = pa_max_int - pa_min_int
print(f"Valor Arredondado - Media: {pa_media_int}")
print(f"Valor Arredondado - Mínimo: {pa_min_int}")
print(f"Valor Arredondado - Máximo: {pa_max_int}")
print(f"Valor Arredondado - Desvio Padrão: {pa_desvio_padrao_int}")
print(f"Valor Arredondado - Amplitude: {pa_amplitude_int}")
print("\n")

# Calcula a distância entre o valor de referência e a aproximação.
erro_absoluto_int = abs(pa_media - pa_media_int)

# Divide o erro absoluto pelo módulo do valor de referência.
erro_relativo_int = erro_absoluto_int / abs(pa_media)

# Converte o erro relativo em porcentagem.
erro_percentual_int = 100.0 * erro_relativo_int


#valores truncados para numero inteiro
pressao_arterial_trunc = np.trunc(pressao_arterial)
pa_media_trunc = np.mean(pressao_arterial_trunc)
pa_min_trunc = min(pressao_arterial_trunc)
pa_max_trunc = max(pressao_arterial_trunc)
pa_desvio_padrao_trunc = np.std(pressao_arterial_trunc)
pa_amplitude_trunc = pa_max_trunc - pa_min_trunc
print(f"Valor Truncado - Media: {pa_media_trunc}")
print(f"Valor Truncado - Mínimo: {pa_min_trunc}")
print(f"Valor Truncado - Máximo: {pa_max_trunc}")
print(f"Valor Truncado - Desvio Padrão: {pa_desvio_padrao_trunc}")
print(f"Valor Truncado - Amplitude: {pa_amplitude_trunc}")

# Calcula a distância entre o valor de referência e a aproximação.
erro_absoluto_trunc = abs(pa_media - pa_media_trunc)

# Divide o erro absoluto pelo módulo do valor de referência.
erro_relativo_trunc = erro_absoluto_trunc / abs(pa_media)

# Converte o erro relativo em porcentagem.
erro_percentual_trunc = 100.0 * erro_relativo_trunc
print("\n")
print(f"Valor Arredondado - Erro Abs: {erro_absoluto_int}")
print(f"Valor Arredondado - Erro Relativo: {erro_relativo_int}")
print(f"Valor Arredondado - Erro %: {erro_percentual_int}")
print("\n")
print(f"Valor Truncado - Erro Abs: {erro_absoluto_trunc}")
print(f"Valor Truncado - Erro Relativo: {erro_relativo_trunc}")
print(f"Valor Truncado - Erro %: {erro_percentual_trunc}")
```
